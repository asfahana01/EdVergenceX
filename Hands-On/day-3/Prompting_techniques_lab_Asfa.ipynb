## 1. Setup & Authentication with Nexus AI

We initialize the OpenAI client pointing to the Nexus AI inference endpoint (`https://nexusapi.navigatelabsai.com`).

```python
import os
import sys
from openai import OpenAI

# Fetch API Key from environment or Colab userdata
api_key = os.environ.get("api_key")
if not api_key:
    try:
        from google.colab import userdata
        api_key = userdata.get("api_key") or userdata.get("api_key")
    except ImportError:
        pass

if not api_key:
    api_key = input("Enter your Nexus API Key: ").strip()

client = OpenAI(
    base_url="https://nexusapi.navigatelabs.ai",
    api_key=api_key
)

print("Nexus AI Prompt Engineering Playground initialized!")
```

## 2. Zero-Shot Prompting

In zero-shot prompting, we give the model a direct instruction without exemplars. We also configure `temperature` and `top_p`.

```python
response_zero = client.chat.completions.create(
    model="nova-micro",
    messages=[
        {"role": "user", "content": "List three distinct benefits of regular aerobic exercise."}
    ],
    temperature=0.7,
    top_p=0.9
)

print("=== Zero-Shot Output ===")
print(response_zero.choices[0].message.content.strip())
```

## 3. Few-Shot Prompting: Tone Transformation

We provide a system instruction and conversational exemplars showing how casual phrases should be rewritten into formal corporate communications.

```python
few_shot_prompt = [
    {"role": "system", "content": "You are a helpful assistant that rewrites sentences to be more formal."},
    {"role": "user", "content": "make it formal: wanna grab lunch?"},
    {"role": "assistant", "content": "Would you like to have lunch together?"},
    {"role": "user", "content": "make it formal: can't come now"},
    {"role": "assistant", "content": "I am unfortunately unable to attend at this time."},
    {"role": "user", "content": "make it formal: shoot me the slides asap"}
]

response_few = client.chat.completions.create(
    model="nova-micro",
    messages=few_shot_prompt,
    temperature=0.0
)

print("=== Few-Shot Formal Rewrite ===")
print(response_few.choices[0].message.content.strip())
```

## 4. Instructional Prompting: Budget Travel Advisor

Here we supply explicit task guidelines, format constraints, and domain scope for a travel itinerary.

```python
response_instruct = client.chat.completions.create(
    model="nova-micro",
    messages=[
        {"role": "system", "content": "You are an expert budget travel advisor specializing in low-cost European itineraries."},
        {"role": "user", "content": "Give me a 3-day itinerary for Paris under a tight budget. Include free attractions, affordable dining, and public transit tips."}
    ],
    temperature=0.3
)

print("=== Instructional Output ===")
print(response_instruct.choices[0].message.content.strip())
```

## 5. Chain-of-Thought (CoT) Prompting: Step-by-Step Reasoning

For mathematical and logical calculations, asking the model to "think step by step" enables intermediate deductive computation before outputting the final result.

```python
cot_prompt = [
    {"role": "user", "content": "A train travels 60 km in 1.5 hours, stops for 30 minutes, and then travels 80 km in 2 hours. What is its average speed for the entire journey in km/h? Think step by step."}
]

response_cot = client.chat.completions.create(
    model="nova-micro",
    messages=cot_prompt,
    temperature=0.0
)

print("=== Chain-of-Thought Reasoning Output ===")
print(response_cot.choices[0].message.content.strip())
```

## 6. Role-Based Prompting: The Shakespearean Poet

We assign a creative persona and stylistic constraint to observe how role-based prompting reshapes explanation style.

```python
role_prompt = [
    {"role": "system", "content": "You are a Shakespearean poet who explains complex scientific concepts strictly in dramatic rhyming verse."},
    {"role": "user", "content": "Explain how photosynthesis works in green plants."}
]

response_role = client.chat.completions.create(
    model="nova-micro",
    messages=role_prompt,
    temperature=0.8
)

print("=== Role-Based Poet Output ===")
print(response_role.choices[0].message.content.strip())
```

## 7. Multi-Model Prompt Playground

On Nexus AI, you can swap models by changing the `model` identifier string. Let us benchmark how different model families respond to the identical prompt.

```python
test_prompt = "Explain the difference between supervised learning and unsupervised learning in 2 short bullet points."

candidate_models = [
    "gpt-4.1-nano",
    "nova-micro"
]

for model_id in candidate_models:
    print(f"\n{'='*20} Testing Model: {model_id} {'='*20}")
    try:
        res = client.chat.completions.create(
            model=model_id,
            messages=[{"role": "user", "content": test_prompt}],
            temperature=0.0
        )
        print(res.choices[0].message.content.strip())
    except Exception as e:
        print(f"Error querying {model_id}: {e}")
```

## 8. Your Turn: Production RTCF Prompt Challenge

**Task:** Build an end-to-end RTCF prompt for an Enterprise Risk Analyst evaluating cloud migration risks.
- **Role:** Principal Cloud Migration Risk Analyst
- **Task:** Identify the top 2 technical risks and top 1 compliance risk for migrating on-premise healthcare records to the cloud
- **Context:** Target audience is hospital board members; HIPAA data privacy compliance is mandatory; assume multi-region failover is required
- **Format:** Markdown table with columns: `[Risk Category, Threat Description, Mitigation Strategy]`
- Execute the request on `gpt-4.1-nano` with `temperature=0.0` and display the formatted report.

```python
rtcf_challenge_prompt = """
### ROLE
You are a Principal Cloud Migration Risk Analyst.

### TASK
Identify the top 2 technical risks and top 1 compliance risk associated
with migrating on-premise healthcare records to the cloud.

### CONTEXT & CONSTRAINTS
- Target audience: Hospital board members with a non-technical background.
- HIPAA data privacy compliance is mandatory.
- Assume multi-region failover is required.
- Focus on the most significant and practical risks.
- Keep threat descriptions clear and concise.
- Provide practical, actionable mitigation strategies.
- Do not include risks outside the requested 2 technical and 1 compliance risks.

### FORMAT
Return the answer exclusively as a Markdown table with exactly these columns:

| Risk Category | Threat Description | Mitigation Strategy |

The table must contain exactly 3 rows:
- 2 rows for Technical Risk
- 1 row for Compliance Risk
"""

response_challenge = client.chat.completions.create(
    model="gpt-4.1-nano",
    messages=[
        {"role": "user", "content": rtcf_challenge_prompt}
    ],
    temperature=0.0
)

print("=== YOUR TURN: RTCF RISK ASSESSMENT TABLE ===")
print(response_challenge.choices[0].message.content.strip())
```
